```python
df = df[["v1", "v2"]]
df.head()
```

```python
df.rename(columns={"v1": "label", "v2": "message"}, inplace=True)
df.head()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df["label"].value_counts()
```

```python
df["message"] = df["message"].str.lower()
df["message"].head()
```

```python
df["length"] = df["message"].apply(len)
df.head()
```

```python
# word tokenisation, stopwords, stemmerization
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
# word tokenisation
w = word_tokenize(x)
```

```python
# listing of stopwords
stop_words = stopwords.words("english")
```

```python
# Remove stopwords
df["message"] = df["message"].apply(
    lambda x: " ".join(
        [word for word in w if word not in stop_words]))
```

```python
# Stemming
p = PorterStemmer()
```

```python
df["message"] = df["message"].apply(
    lambda x: " ".join(
        [p.stem(word) for word in x.split()]))
```

```python
X = df["message"]
Y = df["label"]
```

```python
# TF-IDF Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer

t = TfidfVectorizer(max_features=3000)
X_vec = t.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_vec,Y,test_size=0.20, random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train, Y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

Y_pred = model.predict(X_test)

print(accuracy_score(Y_test, Y_pred), classification_report(Y_test, Y_pred))
```

```python
sample_message = ["free entry win cash prize now"]
sample = t.transform(sample_message)

prediction = model.predict(sample)

print(prediction[0])
```
