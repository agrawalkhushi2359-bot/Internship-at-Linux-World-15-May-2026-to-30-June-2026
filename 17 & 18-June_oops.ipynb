# Encapsulation

```python
class bankaccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        
    def show(self):
        print("name :", self.name)
        print("balance :", self.balance)

        
print("""
    1. enter 1 to enter details
    2. enter 2 to breakout
    """)


while True:
    choice = int(input("enter your choice:"))
    if choice == 1:
        name = input("enter your name :")
        balance = input("enter your balance:")
        b1 = bankaccount(name, balance)       
    else:
        print("you are broked out")
        break
    b1.show()
```

# ClassMethod

```python
class student:
    school = "ABCD"
    @classmethod
    def School(cls):
        print(cls.school)
s = student()
s.School()
```

# StaticMethod

```python
class student:
    @staticmethod
    def School(school):
        print(school)

student.School("ABCD")
```

# Single Level Inheritance

```python
class animal:
    def eat(self):
        print("all animals eats")
class dog(animal):
    def barks(self):
        print("barks")
d = dog()
d.barks()
d.eat()
```

# Multilevel Inheritance

```python
class animal:
    def eat(self):
        print("all animals eats")
class dog(animal):
    def barks(self):
        print("barks")
class puppy(dog):
    def cute(self):
        print("all pupppies are cute")

p = puppy()
p.cute()
p.barks()
p.eat()
```

# Multiple Inheritance

```python
class animal:
    def eat(self):
        print("all animals eats")
class dog():
    def barks(self):
        print("barks")
class puppy(animal, dog):
    def cute(self):
        print("all pupppies are cute")
p = puppy()
p.cute()
p.barks()
p.eat()
```

# Super Keyword

```python
class parent:
    def __init__(self):
        print("parent constructor")

class child(parent):
    def __init__(self):
        super().__init__()                     # calls parent class's constructor
        print("child constructor")

c = child()
```

# Operator Overloading

```python
class addition:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary
        
    def show(self):
        print(self.real,"i","+",self.imaginary, "j")

# without using dender function
    
    def add(self, b):
        self.newReal = self.real + b.real
        self.newImaginary = self.imaginary + b.imaginary
        print(self.newReal,"i","+",self.newImaginary, "j")

# using dender function
    
    def __add__(self, b):
        self.newReal = self.real + b.real
        self.newImaginary = self.imaginary + b.imaginary
        print(self.newReal,"i","+",self.newImaginary, "j")


a = addition(5, 3)
b  = addition(6, 4)
a.show()
b.show()
a.add(b)
a+b
```

# setter method & getter method

```python
class Student:   
    def __init__(self):
        self.name = "KA"
        self.score = 50   # Protected attribute
# Getter
    @property
    def marks(self):
        print("score is:",self.score)
# Setter
    @marks.setter
    def marks(self, value):
        self.score = value
s1 = Student()
print("Name:", s1.name)
print("Marks:", s1.score)
# Getter call
print(s1.marks)     # getter function marks() is calling internally
# Setter call
s1.marks = 95
print(s1.marks)
# Invalid value
s1.marks = 120
```
